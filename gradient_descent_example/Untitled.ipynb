```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

# Data generation (same as before)
np.random.seed(0)
m = 100
age = np.random.uniform(18, 70, m)
income = np.random.uniform(20, 100, m)
labels = (age > 35) & (income > 50)
y_train = labels.astype(int)
X_train = np.c_[age, income]  # No manual bias term needed

# Normalize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Train logistic regression model
model = LogisticRegression(solver='lbfgs', max_iter=15000)  # lbfgs is a robust solver
model.fit(X_train_scaled, y_train)

# Print learned parameters (intercept is bias, coef_ are weights)
print(f"Intercept (θ0): {model.intercept_[0]:.4f}")
print(f"Coefficients (θ1, θ2): {model.coef_[0]}")

# Prediction function
def get_probability(age, income):
    X_test = np.array([[age, income]])
    X_test_scaled = scaler.transform(X_test)
    return model.predict_proba(X_test_scaled)[0, 1]  # Probability of class 1

# Test prediction
test_age = 27
test_income = 27
prob = get_probability(test_age, test_income)
print(f"Age = {test_age}, Income = {test_income}, Probability = {prob:.4f}")

# Find ~50% probability point (e.g., income = 50, solve for age)
test_income = 50
for age in [58, 59]:
    prob = get_probability(age, test_income)
    print(f"Age = {age}, Income = {test_income}, Probability = {prob:.4f}")
```
